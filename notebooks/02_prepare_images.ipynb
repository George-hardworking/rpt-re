```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

from config import IMAGES_ROOT, OHLC_PARQUET, WINDOW_DAYS
from viz.preview_images import (
    build_sample_windows,
    default_permno,
    memmap_image_path,
    plot_memmap_index,
    plot_sample_windows,
)
```

```python
import pandas as pd

from data.calendar import as_of_dates_for_window, market_calendar
from data.images import image_shape

calendar = market_calendar(pd.read_parquet(OHLC_PARQUET, columns=["DlyCalDt"])["DlyCalDt"])
for w in WINDOW_DAYS:
    as_ofs = as_of_dates_for_window(w, calendar)
    h, width = image_shape(w)
    print(f"window={w}d  shape={h}x{width}  as_of={len(as_ofs)}  {as_ofs[0].date()} .. {as_ofs[-1].date()}")
```

```python
permno = default_permno(OHLC_PARQUET)
samples = build_sample_windows(OHLC_PARQUET, permno=permno, window_days_list=WINDOW_DAYS)
for window_days, _img, label, as_of in samples:
    print(f"I{window_days} as_of={as_of.date()}  {label}")
plot_sample_windows(samples, permno)
```
